## <span style="color:red">Class</span>

A **class** is a blueprint or template used to create objects in object-oriented programming (OOP). It defines the properties (variables) and behaviors (functions/methods) that objects will have.

```python
class Student:
    def __init__(self,name):
        self.name = name
s1 = Student("Aadityendra")
print(s1.name)
```

```python
class Car:
    color = "White"
    Price = "$250000"
    Seat = "2"
s1 = Car()
print(s1.color)
print(s1.Price)
print(s1.Seat)
    
```

```python
class Students:
    name = "Aadityendra"
    roll = 1
    branch = "AI & DS"
s1 = Students()
s2 = Students()

print(s1.name)
print(s2.roll)
print(s1.branch)
```

```python
class College:
    college_name= "Engineering College Bikaner"
    college_branch = "AI & DS"
    college_address = "Karni Inducustrial Area"
    college_city = "Bikaner"
s1 = College()
s2 = College()
s3 = College()
s4 = College()

print(s1.college_name)
print(s2.college_branch)
print(s3.college_city)
print(s4.college_address)
```

```python
class College:
    def show(self):             # Mandatory to pass self argument in a function
        print("hello")
s1 = College()
s1.show()
```

```python
class Parent:
    name = "Aadi"
    age = 18
    height = 6
    def show(self):
        print(self.name)
        print(self.age)
        print(self.height)
c1 = Parent()
c1.show()
```
