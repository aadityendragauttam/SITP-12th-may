### Conditional statement
A **conditional statement** in Python is a statement that allows a program to make decisions and execute different blocks of code based on whether a condition is *True* or *False*.

**<span style="color:cyan">Types of Conditional statements : </span>**
* if
* else
* elif 

```python
a = int(input("Enter a number : "))
b = int(input("Enter another number : "))
sum = a + b
print(sum)
print(type(sum))
```

```python
balance = 5000
withdraw = int(input("Enter Withdrawl amount :"))
if withdraw <= balance :
    print("Transfered successfully")
    total = balance - withdraw
    print("Balance left :",total)
else :
    print("Insufficient Balance!!")
```

```python
marks = int(input("Enter your marks : "))
if marks >= 36:
    print("Pass")
else:
    print("Fail")
```

```python
light = input("Enter Signal color :")
if  light.lower() == "red":
    print("Stop")
elif light.lower() == "yellow":
    print("Ready")
elif light.lower() == "green":
    print("Go")
else :
    print("Invalid color")
```

```python
age = int(input("Enter your age : "))
if age>=18:
    print("You can vote")
else:
    print("You can't vote")
```

```python
amount = int(input("Enter your amount : "))
free = 200
if amount >=200:
    print("YAY!! You recieved a free delivery")   
else:
    new = free -amount
    print("Select product worth",new ,"to get a free delivery ")
```

```python
marks = int(input("Enter marks of the students : "))
if(marks>=90):
    print("Grade A")
elif(marks<90 and marks>=80):
    print("Grade B")
elif(marks<80 and marks>=70):
    print("Grade C")
elif(marks<70 and marks>=60):
    print("Grade D")
elif(marks<60 and marks>=50):
    print("Grade E")
elif(marks<50 and marks>=40):
    print("Grade E+")
else:
    print("F")
```
