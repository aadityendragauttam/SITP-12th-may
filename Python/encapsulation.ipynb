## <span style ="color:red">**Encapsulation :**</span>
Encapsulation is binding of data Members(variables) and methods (functions) inside a class and restricting direct access to some data.


1. Public Members: Public members can be accessed from anywhere.
2. Private Members: They are meant to be accessed only inside the class.Private members use double underscores __.
3. Protected Members:They are intended to be accessed only inside the class and its subclasses.Protected members use a single underscore _.


```python
class Parent:
    name = "Aadityendra"
    _age = 18
    __height = 6
    def show(self):
        print(self.__height)
```

```python
s1 = Parent()
```

```python
print(s1.name)
```

```python
print(s1._age)
```

```python
#print(s1.__height)           #error
```

### Method to access Private variable:

```python
print(s1._Parent__height)
```

```python
s1.show()
```
