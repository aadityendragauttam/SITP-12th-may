### Exception Handling
`Exception handling` in Python is the mechanism used to detect and respond to runtime errors without abruptly terminating the program.

How it works
* try: Contains code that might cause an exception.
* except: Handles specific exceptions.
* else: Executes if no exception occurs.
* finally: Always executes, whether an exception occurs or not.

### Common Exception
| Exception           | Description                        |
| ------------------- | ---------------------------------- |
| `ValueError`        | Invalid value for an operation     |
| `TypeError`         | Operation on an inappropriate type |
| `ZeroDivisionError` | Division by zero                   |
| `IndexError`        | List index out of range            |
| `KeyError`          | Dictionary key not found           |
| `FileNotFoundError` | File does not exist                |
| `AttributeError`    | Object has no requested attribute  |


```python
try:
    a = int(input("Enter First number : "))
    b = int(input("Enter Second number : "))
    print(a/b)
except ZeroDivisionError :
    print(f"Cannot divide {a} by zero")
except ValueError:
    print(f"Can't divide int with str ")
```

```python
try:
    a = [1,2,3,47]
    ind = int(input("Enter index of list to be access : "))
    print(a[ind])
except ValueError:
    print("Enter a valid input")
except IndexError:
    print("Index doesnt exists")
```

```python
try: 
    age = int(input("Enter age: "))
    if age<0:
        raise Exception("AGE must be positive")
    else:
        print(age)
except ValueError:
    print("Invalid input")
```

```python
class InvalidAgeError(Exception):
    pass
try:
    age = int(input("Enter your age : "))
    if age<18:  
        raise InvalidAgeError("You aren't eligible to Vote")
    else:
        print("You are eligible to vote")
except Exception as e:
    print(e)
```

```python
class Kuchbhi(Exception):
    pass
try:
    amount = int(input("Enter Amount to be withdrawl"))
    balance = 7000
    if amount > balance:
        raise Kuchbhi("Gareeebbbbbb")
    else:
        print("Withdrawl Successfull, Balance left : ",balance - amount)
except Exception  as e:
    print(e)
```

```python
class PinlenError(Exception):
    pass
try:
    pin = int(input('Enter Your pin'))
    if len(pin)>4:
        raise PinlenError("Pin lenght must be 4")
    else:
        print("Pin Excepted")
except Exception as e:
    print(e)

```
