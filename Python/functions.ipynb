#  <span style="color:red">Functions: </span>
**Function** is a reusable block of code designed to perform a specific task. Functions help organize programs, reduce repetition, and make code easier to understand.

* print()
* range()
* type()

* append()

* casefold()

### <span style="color:white"> User defined functions:</span>
* `def` keyword is used.


```python
def hello():
    print("hello")
hello()
```

```python
def sum(a,b):
    return a+b
print(sum(2,3))
```

```python
def square(x):
    print(x**2)
square(3)
```

```python
def even(lis):
    print("Even items: ")
    for x in lis:
        if x%2 == 0:
            print(x)
def odd(lis):
    print("Odd items: ")
    for x in lis:
        if x%2!= 0:
            print(x)
lis = [1,2,3,4,5,5,6,6,4,342,55,536,3453]
even(lis)
odd(lis)
```

```python
def prime(x):
    if x>1:
        for i in range(2,x):
            if x%i == 0:
                print("Not Prime")
                break
        else:
            print("Prime")
    else:
        print("Not prime")
prime(567675986756453)
```
