Q1.) Create a class `NumberSeries` that uses a loop to print numbers from 1 to 100. 

```python
class NumberSeries:
    def print_num(self):
        for x in range(1,101):
            print(x)

s1 = NumberSeries()
s1.print_num()
```

Q2.) Create a class `EvenOddChecker` that checks and prints whether numbers from 1 to 50 are even or odd using a loop. 

```python
class EvenOddChecker:
    def show(self):
        for x in range(1,51):
            if x % 2 == 0:
                print(f"{x} is even")
            else:
                print(f"{x} is odd")
s1 = EvenOddChecker()
s1.show()
```

Q3.) Create a class `MultiplicationTable` with a function to print the multiplication table of a user-given number. 

```python
class MultiplicationTable:
    def mul(self,num):
        for x in range(1,11):
            print(x * num)
            
s1 = MultiplicationTable()
num = int(input("Enter a Number : "))
s1.mul(num)


```

Q4.) Create a class `FactorialCalculator` with a function to calculate the factorial of a number using a loop. 

```python
class FactorialCalculator:
    def factorial(self,num):
        fact = 1
        for x in range(1,num+1):
            fact = fact*x
        print(fact)
s1 = FactorialCalculator()
s1.factorial(5)

            
```

Q5.) Create a class `PrimeNumberChecker` with a function to check whether a number is prime or not. 

```python
class PrimeNumberChecker:
    def prime(self,num):
        if num > 0:
            for x in range(2,num):
                if num % x == 0:
                    print(f"{num} is not prime number")
                    break
            else:
                print(f"{num} is a prime number")
        else:
            print(f"{num} is not prime")
s1 = PrimeNumberChecker()
num = int(input("Enter a number: "))
s1.prime(num)
```

Q8.) Create a class `Employee` using a constructor **(__ init __)** to store employee details and create a function to calculate yearly salary

```python
class Employee:
    def __init__(self,name,role,monthly_salary):
        self.name = name
        self.role = role
        self.monthly_salary = monthly_salary
    def yearly_salary(self):
        print("Yearly Salary : " ,self.monthly_salary * 12)
s1 = Employee("Aadityendra","Data Scientist",150000)
print(s1.name)
print(s1.role)
print(s1.monthly_salary)
s1.yearly_salary()
```

Q10.) Create three classes `Vehicle`, `Car`, and `SportsCar` using **multilevel inheritance** and display information of a sports car

```python
class Vehicle:
    brand = "Koenigsegg"
    def vehicle_info(self):
        print(f"Vehicle Brand : {self.brand}")
class Car(Vehicle):
    model = "Jesko"
    def car_info(self):
        print(f"Car Model : {self.model}")
class SportsCar(Car):
    top_speed ="563 kmph"
    def sc_info(self):
        print(f"Top Speed : {self.top_speed}")
s1 = SportsCar()
s1.vehicle_info()
s1.car_info()
s1.sc_info()
```

Q11.)  Create a class `BankAccount` with functions: 

* deposit money

* withdraw money 

* check balance 


```python
class BankAccount:
    balance = 5000
    def deposite(self,deposite_amount):
        if deposite_amount > 0:
            self.balance += deposite_amount
            print("Money deposited Successfully")
        else:
            print("Invalid Amount")
    def withdraw(self,withdraw_amount):
        if 0 < withdraw_amount < self.balance:
            print("Money withdraw Successfully")
            self.balance -= withdraw_amount
        else:
            print("Insuffient balance or Invalid amount")
    def balance_info(self):
        print(f"Total balance : {self.balance}")
s1 = BankAccount()
s1.deposite(400)
s1.withdraw(300)
s1.balance_info()
```
