## <span style ="color:red">**Inheritance**</span>

`Inheritance` in Python is a fundamental concept of object-oriented programming (OOP) where a class (child/subclass) acquires the properties and methods of another class (parent/superclass).



| Type of Inheritance          | Description                                                        | Example Syntax                                                                |
| ---------------------------- | ------------------------------------------------------------------ | ----------------------------------------------------------------------------- |
| **Single Inheritance**       | One child class inherits from one parent class.                    | `python class A: pass class B(A): pass `                                      |
| **Multiple Inheritance**     | One child class inherits from multiple parent classes.             | `python class A: pass class B: pass class C(A, B): pass `                     |
| **Multilevel Inheritance**   | A class inherits from a class that already inherits another class. | `python class A: pass class B(A): pass class C(B): pass `                     |
| **Hierarchical Inheritance** | Multiple child classes inherit from the same parent class.         | `python class A: pass class B(A): pass class C(A): pass `                     |
| **Hybrid Inheritance**       | Combination of two or more types of inheritance.                   | `python class A: pass class B(A): pass class C(A): pass class D(B, C): pass ` |


### **1). Single Inheritance**

```python
class Parent:
    name = "Aadi"
    age = 18
    height = 6
    def show(self):
        print(self.name)
        print(self.age)
        print(self.height)
class Child(Parent):
    roll_no = 1
c1 = Child()
c1.show()
print(c1.roll_no)
```

### **2.) Multilevel Inheritance**

```python
class Level1:
    name = "Aadi"
    age = 18
    height = 6
    def display(self):
        print(self.name)
        print(self.age)
        print(self.height)
class Level2(Level1):
    roll_no = 1
class Level3(Level2):
    pass
c1 = Level3()
c1.display()
print(c1.roll_no)
```

### **3.) Multiple Inheritance**

```python
class Subject1:
    def subject1_info(self):
        print("I am Subject 1 Class method")
class Subject2:
    def subject2_info(self):
        print("I am Subject 2 Class method")
class Students(Subject1,Subject2):
    def student_info(self):
        print("I am Student Class method")
    pass
h = Students()
h.subject1_info()
h.subject2_info()
```

### **4.) Hybrid Inheritance:**

```python
class Student:
    def student_info(self):
        print("I am Student class method ")
class Subject1(Student):
    def subject1_info(self):
        print("I am Subject1 class method")
class Subject2(Student):
    def subject2_info(self):
        print("I am Subject2 class method")  
class Teacher(Subject1,Subject2):
    def teacher_info(self):
        print("I am Teacher class method")
s1 = Teacher()
s1.student_info()
s1.subject1_info()
s1.subject2_info()
s1.teacher_info()
```

### **5.) Hierarchical Inheritance**

```python
class Student:
    def student_info(self):
        print("I am Student class method ")
class Subject1(Student):
    def subject1_info(self):
        print("I am Subject1 class method")
class Subject2(Student):
    def subject2_info(self):
        print("I am Subject2 class method")  
s1 = Subject1()
s2 = Subject2()
s1.student_info()
s2.student_info()
```
