### Loops in python
A **loop** in Python is a programming statement used to repeat a block of code multiple times until a condition is met or for each item in a sequence.

**<span style="color:cyan">Type of loops in python</span>**
* For Loop
* While Loop

```python
lst = [1,2,3,4,5,6]
for x in lst:
    print(x)
```

```python
name  = "Aadityendra"
for x in name:
    print(x)
```

```python
for x in range(1,10,2):
    print(x)
```

```python
Student = {
    "name":"Aadityendra",
    "age":18,
    "Roll no.":1,
    "Branch":"AI & DS"
}
for x in Student:
    print(x,":",Student[x])
print(Student.items())
```

```python
lst = [1,2,3,4,5,4,3,2,1]
for val in lst:
    if val == 2:
        continue            #skip
    if val == 5:
        break               #terminates the loop
    print (val)
```

```python
fruits =["banana","mango","grapes","watermelon"]
for val in fruits:
    if val == "mango":
        continue
    print(val)
```

```python
for x in range(6):
    if x == 2:
        continue
    print(x)
```

```python
lst = [1,2,['h','e','l','l','o'],3.431]
print(lst[2][1])
```

### <span style="color:cyan"> While Loop: </span>
`while loop` is a programming control structure that repeatedly executes a block of code as long as a condition is **true**.


```python
i = 1
while i<=5:
    print(i)
    i+=1
```

```python
i=5
while i>=1:
    print(i)
    i-=1
```
