## <span style ="color:red"> **Constructor :** </span>
Constructor is a special method used to initialize an object when a class is created.

The constructor method in Python is called **__ init __()**.
Initialization :
* class ClassName:

        def __init__(self, parameters):

            # initialization code

```python
class Constructor:
    def __init__(self,name,age,height):
        self.name = name
        self.age = age
        self.height = height
s1 = Constructor("Aadityendra",18,6)
print(s1.name)
print(s1.age)
print(s1.height)
```

```python
class College:
    def __init__(self):
        self.name = "Aadityendra"
s1 = College()
print(s1.name)
```

```python
class Constructor:
    def __init__(self,name,age,height):
        self.name = name
        self.age = age
        self.height = height
    def show(self):
        print(self.name)
s1 = Constructor("Aadityendra",18,6)
s1.show()
print(s1.age)
print(s1.height)
```

```python
class Employee:
    def __init__(self, name):
        self.name = name
        print("Employee Constructor Called")


class Developer:
    def __init__(self, language):
        self.language = language
        print("Developer Constructor Called")


class Leader(Employee, Developer):
    def __init__(self, name, language, team_size):
        Employee.__init__(self, name)
        Developer.__init__(self, language)
        self.team_size = team_size
        print("Leader Constructor Called")


t1 = Leader("Aadityendra", "Python", 8)

print("Name :", t1.name)
print("Programming Language :", t1.language)
print("Team Size :", t1.team_size)
```
