Q1.) Create a class `Employee` with attributes emp_name, emp_id, and salary. Create 3 objects and display their details. 

```python
class Employee:
    def __init__(self,emp_name,emp_id,salary):
        self.emp_name = emp_name
        self.emp_id = emp_id
        self.salary = salary
s1 = Employee('Aadityendra',123456789,150000)
print(s1.emp_name)
print(s1.emp_id)
print(s1.salary)
```

Q2.) Create a class Calculator with methods for addition, subtraction, multiplication, and division. 

```python
class Calculator:
    def __init__(self,a,b):
        self.a = a
        self.b = b
    def addition(self):
        print("Sum : ",self.a + self.b)
    def subtraction(self):
        print("Subtraction : ",self.a - self.b)
    def multiplication(self):
        print("Multiplication : ",self.a * self.b)
    def division(self):
        quotient = self.a / self.b
        remainder = self.a % self.b
        print(f"Division : \n Quotient : {quotient}\n Remainder : {remainder}")
s1 = Calculator(10,2)
s1.addition()
s1.subtraction()
s1.multiplication()
s1.division()
```

Q3.) Create a class Book with a constructor to initialize title, author, and price, then print details. 

```python
class Book:
    def __init__(self,title,author,price):
        self.title = title
        self.author = author
        self.price = price
s1 = Book('The Invisible man','H.G wells',114)
print(f" Book name : {s1.title} \n Author : {s1.author} \n Price : {s1.price}")
```

Q5.) Write a Python program using regex to find all digits from the string: "Python123Assignment456" 

```python
import re
text = "Python123Assignment456"
digits = re.findall(r"\d+", text)
print("Extracted digits:", digits)
```
