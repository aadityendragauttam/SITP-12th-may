Q2.) Create a class `PrimeChecker` with a function to check whether a number is prime or not.

```python
class PrimeNumberChecker:
    def __init__(self,num):
        self.num = num
    def prime(self):
        if self.num > 0:
            for x in range(2,self.num):
                if self.num % x == 0:
                    print(f"{self.num} is not prime number")
                    break
            else:
                print(f"{self.num} is a prime number")
        else:
            print(f"{self.num} is not prime")
s1 = PrimeNumberChecker(2)
s1.prime()
```

Q5.) Create a class Calculator with functions: 
* add() 
* subtract() 
* multiply() 
* divide() 

```python
class Calculator:
    def __init__(self,a,b):
        self.a = a
        self.b = b
    def add(self):
        print("Sum : ",self.a + self.b)
    def subtract(self):
        print("Subtraction : ",self.a - self.b)
    def multiply(self):
        print("Multiplication : ",self.a * self.b)
    def divide(self):
        quotient = self.a / self.b
        remainder = self.a % self.b
        print(f"Division : \n Quotient : {quotient}\n Remainder : {remainder}")
s1 = Calculator(10,2)
s1.add()
s1.subtract()
s1.multiply()
s1.divide()
```

Q6.) Create a parent class `Person` and child class `Employee` using single inheritance. Display employee details. 

```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def person_info(self):
        print("Name = ",self.name)
        print("Age = ",self.age)
class Employee(Person):
    role = "Developer"
    def employee_info(self):
        print("Role = ",self.role)
s1 = Employee("Aadityendra",18)
s1.person_info()
s1.employee_info()
```

Q7.) Create a class BankAccount with functions: 
* deposit() 
* withdraw() 
* check_balance()

```python
class BankAccount:
    def __init__(self,balance,deposite_amount,withdraw_amount):
        self.balance = balance
        self.deposite_amount = deposite_amount
        self.withdraw_amount = withdraw_amount
    def deposite(self):
        if self.deposite_amount > 0:
            self.balance += self.deposite_amount
            print("Money deposited Successfully")
        else:
            print("Invalid Amount")
    def withdraw(self):
        if 0 < self.withdraw_amount < self.balance:
            print("Money withdraw Successfully")
            self.balance -= self.withdraw_amount
        else:
            print("Insuffient balance or Invalid amount")
    def balance_info(self):
        print(f"Total balance : {self.balance}")
s1 = BankAccount(5000,200,150)
s1.deposite()
s1.withdraw()
s1.balance_info()
```

Q8.) Create a class `FactorialCalculator` with a function to calculate factorial of a number using a loop. 

```python
class FactorialCalculator:
    def __init__(self,num):
        self.num = num
    def factorial(self):
        fact = 1
        for x in range(1,self.num + 1):
            fact = fact*x
        print(fact)
s1 = FactorialCalculator(5)
s1.factorial()
```
