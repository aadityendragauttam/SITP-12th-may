```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4]
y = [2,4,6,8]
plt.plot(x,y,'o')               #to plot the graph
plt.show()
```

```python
x = [1,2,3,4]
y = [2,4,6,8]
plt.plot(x,y,marker='*')               #to plot the graph
plt.show()
```

```python
x = [1,2,3,4]
y = [2,4,6,8]
plt.plot(x,y,'o:r')               #o = circle, : = dotted line, r = red colour
plt.show()
```

```python
x = [1,2,3,4]
y = [2,4,6,8]
plt.plot(x,y,'o-r')               #o = circle, : = dotted line, r = red colour
plt.show()
```

```python
x = [1,2,3,4]
y = [2,4,6,8]
plt.plot(x,y,ms = 20,marker = 'o',color = 'red')               #ms = marker size
# returns address of plotted graph
```

```python
x = [1,2,3,4]
y = [2,4,6,8]
plt.plot(x,y,ms = 20 ,marker= "*")              
plt.show()
```

```python
x = [1,2,3,4]
y = [2,4,6,8]
plt.plot(x,y,ms = 20 ,marker= "*",mfc = 'red')               #mfc = andar ka color change krsi marker ke
plt.show()
```

```python
x = [1,2,3,4]
y = [2,4,6,8]
plt.plot(x,y,ms = 20 ,marker= "*",mfc = 'red',mec = "yellow")               #mec = border ka color change krsi marker ke
plt.show()
```

```python
arr =[1,3,5,7]
plt.plot(arr,ls= 'dashed')
plt.show()
```

```python
arr =[1,3,5,7]
plt.plot(arr,ls= 'dotted')
plt.show()
```

```python
arr =[1,3,5,7]
plt.plot(arr,linewidth = 20)
plt.show()
```

```python
x = [1,3,5,7]
y = [4,5,6,7]
plt.plot(x,color = 'red')
plt.plot(y)
plt.show()
```

```python
x = [1,3,5,7]
y = [4,5,6,7]
plt.xlabel("oiuytf")
plt.ylabel('yaxis')
plt.title('Matplotlib')
plt.plot(x)
plt.show()
```

```python
x = [1,3,5,7]
y = [4,5,6,7]
plt.xlabel("oiuytf")
plt.ylabel('yaxis')
plt.title('Matplotlib')
plt.grid()
plt.plot(x)
plt.show()
```

```python
plt.scatter(x,y)
plt.show()
```

```python
plt.hist(x,y)
plt.show()
```

```python
plt.bar(x,y)
plt.show()
```

```python
plt.pie(x)
plt.show()
```
