### File Handling
File handling in Python means working with files — creating, opening, reading, writing, appending, and closing files using Python programs.
| Mode  | Meaning                         |
| ----- | ------------------------------- |
| `"r"` | Read file                       |
| `"w"` | Write file (creates/overwrites) |
| `"a"` | Append data                     |
| `"x"` | Create new file                 |

```python
file = open("hello.txt",'r')
```

```python
file.read()
```

```python
# file.readline()
# file.readlines()
```

```python
file.close()
```

```python
with open('hello.txt') as file:
    print(file.read())
```

```python
file = open('hello.txt','w')
```

```python
file.write("Hii How are you")
```

```python
file.close()
```

```python
file = open('autofile.txt','w')         #automatically create a file named autofile.txt
file.write("Modiiiiiiiiiii")
file.close()
file = open('autofile.txt','r')
file.read()
```

```python
file.close()
```

```python
file = open("autofile.txt",'a')
file.write(" ,Phle modiji aaye phir unhone brahmand banaya")
file.close()
```
